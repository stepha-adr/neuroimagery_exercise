## Partie 0 – Rappels

> **Rappel – forme et nombre d'éléments.**
> Un tableau de forme `(a, b, c)` contient `a * b * c` nombres. `.ndim` donne le nombre de dimensions, `.size` le nombre total d'éléments.
> `reshape` peut donner n'importe quelle nouvelle forme **à condition que le nombre total d'éléments reste le même**. Une des dimensions peut s'écrire `-1` : NumPy la calcule pour toi.

> **Rappel – l'ordre du reshape.**
> `reshape` lit les nombres dans l'ordre, l'axe **le plus à droite** changeant le plus vite (comme pour lire un livre : le long d'une ligne, puis la ligne suivante, puis la page suivante). Refaire un reshape vers la forme d'origine redonne toujours le tableau d'origine.

````python
a = np.arange(12)            # 0, 1, ..., 11
print(a.shape)
b = a.reshape(3, 4)          # 3 lignes de 4
print(b)
print(b.reshape(-1))         # retour en 1D : même ordre
print(b.reshape(2, -1).shape)  # NumPy calcule -1 = 6
````

> **Rappel – les boucles `for`.**
> `for i in range(n):` répète le bloc indenté pour `i = 0, 1, …, n-1`.
> Pour garder un résultat par itération, on peut soit l'ajouter à une liste, soit créer d'abord un tableau de la bonne taille (`np.zeros(n)`) et le remplir.

````python
carres = []
for i in range(5):
    carres.append(i ** 2)
print(carres)

cubes = np.zeros(5)
for i in range(5):
    cubes[i] = i ** 3
print(cubes)
````
